```python
# Imports and whatnot
import os
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import keras
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten
from keras.layers import Conv2D, MaxPooling2D
```

```python
inputshape = (6, 5, 150)
# timeframe, features, window

def compile_model(model):
    # create the model
    # compile the model
    model.compile(loss=keras.losses.mean_squared_error,
                  optimizer=keras.optimizers.Adam(amsgrad=True),
                  # For the metrics, always have AT LEAST these two
                  metrics=["MeanAbsolutePercentageError", "MeanSquaredError"])
    return model
```

```python
def dingus3():
    inputs = keras.Input(shape=inputshape)
    x = Flatten()(inputs)

    x = Dense(6000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(6000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(4000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(4000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(1000, activation='silu')(x)
    x = Dense(5000, activation='silu')(x)

    outputminmax = Dense(2, name = "minmaxprediction")(x)
    outputerrors = Dense(2, activation='relu', name = "errorprediction")(x)

    outputs = keras.layers.concatenate([outputminmax, outputerrors])

    return keras.Model(inputs=inputs, outputs=outputs)
```

```python
def dingus4():
    inputs = keras.Input(shape=inputshape)
    x = Flatten()(inputs)

    x = Dense(6000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(6000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(5000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(4000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(4000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(1000, activation='silu')(x)

    minmax = Dense(2, name = "minmax")(x)
    errors = Dense(2, activation='relu', name = "error")(x)
    decision = Dense(3, activation="softmax", name="decision")(x)

    outputs = keras.layers.concatenate([minmax, errors, decision])

    return keras.Model(inputs=inputs, outputs=outputs)
```

```python
def dingus6():
    inputs = keras.Input(shape=inputshape)
    x = Flatten()(inputs)

    x = Dense(6000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(6000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(5000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(5000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(4000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(4000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(1000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(1000, activation='silu')(x)
    x = Dropout(0.1)(x)
    x = Dense(1000, activation='silu')(x)

    minmax = Dense(2, name = "minmax")(x)
    errors = Dense(2, activation='relu', name = "error")(x)
    decision = Dense(3, activation="softmax", name="decision")(x)

    outputs = keras.layers.concatenate([minmax, errors, decision])

    return keras.Model(inputs=inputs, outputs=outputs)
```

```python
Model_Name = "dingus6"

model = compile_model(dingus4())
print("Model Compiled")
model.summary()
```
